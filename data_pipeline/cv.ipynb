```python
all_b=[]
for i in uls:
    response = requests.get(i)
    fnd = BeautifulSoup(response.text,"html.parser")
    names = fnd.find_all("h3")
    rating = fnd.find_all("p",class_="star-rating")
    price = fnd.find_all("p",class_="price_color")
    stock = fnd.find_all("p",class_="instock availability")
    cat = fnd.find("div",class_="page-header action")
    ns = []
    for i in stock:
        ns.append(i.text.strip())
    new_r = []
    for i in rating:
        new_r.append(i["class"][1])
    new_p = []
    for i in price:
        x = float(i.text[2:])
        new_p.append(float(f'{x*105.50:.2f}'))
    stint = {"In stock":1,}
    sn = [stint.get(s,0) for s in ns]
    x = []
    fx = []
    dsp = {"One":1,"Two":2,"Three":3,"Four":4,"Five":5}
    csd = [dsp.get(item,0) for item in new_r]
    for n,np,p,r,s in zip(names,new_p,price,csd,sn):
        rf = f"{n.text.strip()},{np},{np/105.50:.2f},{r},{s}"
        fx.append(rf)
    for n,r,p,s in zip(names,csd,new_p,stock):
        ds = n.text.strip(),r,p,s.text.strip(),cat.text.strip()
        x.append(ds)
    all_b.extend(x)
fd = pd.DataFrame(all_b,columns=["book name","book rating","book price","book stock","category"])
fd.to_csv("cv.csv",index=False,mode='w')
print(rf)
```
